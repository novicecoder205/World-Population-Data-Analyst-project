```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("world_population_data.csv")
```

```python
df.head(10)
```

```python
df.columns
```

```python
df.describe()
```

```python
# Check to see if there are any NA values in the data.
df.isna().sum()
```

```python
#Check for duplicate value
df.duplicated().sum()
```

```python
# Generalization: Visualize distribution of 2023 population.
sns.histplot(df["2023 population"], bins = 10, kde = True)
plt.title("Distribution of 2023 Population")
plt.xlabel("Population")
plt.ylabel("Frequency")
plt.show()
```

```python
# Group data by continent and calculate the mean population for each continents.
df.groupby("continent")["2023 population"].mean()
```

```python
df.groupby("continent")["2023 population"].mean().sort_values(ascending = False).head(10)
```

```python
df.sort_values(by="2023 population", ascending = True)
```

```python
df.sort_values(by="2023 population", ascending = False).head(10)
```

```python
to_10_countries = df.sort_values(by="2023 population", ascending = False).head(10)
```

```python
sns.barplot(x="2023 population", y = "country", data = to_10_countries)
plt.title("top 10 Countries by Population in 2023")
plt.xlabel("Population")
plt.ylabel("Country")
plt.show()
```

```python
df.columns
```

```python
df.rename(columns = {"density (km²)":"density (km2)"},inplace = True)
```

```python
sns.scatterplot(x="2023 population", y = "density (km2)", data = df)
plt.title("Population and Density Scatter plot")
plt.xlabel("2023 Population")
plt.ylabel("Density")
plt.show()
```

```python
df.columns
```

```python
df.rename(columns = {"area (km²)":"area (km2)"},inplace = True)
```

```python
df.groupby("continent")["area (km2)"].sum().sort_values(ascending = False)
```

```python
sns.violinplot(x="continent", y = "density (km2)", data = df, hue = "continent", palette="viridis")
plt.title("Population Density Distribution Across Continents")
plt.xlabel("Continent")
plt.ylabel("Population Density")
plt.show()
```

```python
sns.boxplot(x="continent", y = "density (km2)", data = df, hue = "continent", palette="muted")
plt.title("Population Density Distribution Across Continents (Without Outliers)")
plt.xlabel("Continent")
plt.ylabel("Population Density")
plt.show()
```

```python
sns.boxplot(x="continent", y = "density (km2)", data = df, hue = "continent", palette="muted", showfliers = False)
plt.title("Population Density Distribution Across Continents (Without Outliers)")
plt.xlabel("Continent")
plt.ylabel("Population Density")
plt.show()
```
